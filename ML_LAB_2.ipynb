```python
import pandas as pd
df = pd.read_excel("Lab Session Data.xlsx", sheet_name="marketing_campaign")
def label_encoding(df, column):
    unique = []
    encoding = {}
    for value in df[column]:
        if value not in unique:
            unique.append(value)
    for i in range(len(unique)):
        encoding[unique[i]] = i

    for i in range(len(df)):
        df.at[i, column] = encoding[df.at[i, column]]

    return df
def one_hot_encoding(df, column):
    unique = []
    for value in df[column]:
        if value not in unique:
            unique.append(value)
    matrix = []
    for value in df[column]:
        row = []
        for category in unique:
            if value == category:
                row.append(1)
            else:
                row.append(0)
        matrix.append(row)
    for j in range(len(unique)):
        values = []
        for i in range(len(matrix)):
            values.append(matrix[i][j])
        df[unique[j]] = values
    df.drop(column, axis=1, inplace=True)
    return df
print("Original Dataset Shape:", df.shape)
df = label_encoding(df, "Education")
df = one_hot_encoding(df, "Marital_Status")
print("Dataset Shape After Encoding:", df.shape)

print(df.head())
```

```python
''''a4 and a5'''
def minkowski_distance(A, B, p):
    distance = 0
    for i in range(len(A)):
        distance = distance + abs(A[i] - B[i]) ** p
    return distance ** (1 / p)
A = []
B = []
A.append(df["Income"][0])
A.append(df["Kidhome"][0])
A.append(df["Teenhome"][0])
B.append(df["Income"][1])
B.append(df["Kidhome"][1])
B.append(df["Teenhome"][1])

print("Manhattan Distance =", minkowski_distance(A, B, 1))
print("Euclidean Distance =", minkowski_distance(A, B, 2))
```

```python
'''a6'''
from scipy.spatial.distance import minkowski
print("p\tMy Function\tScipy")
for p in range(1, 11):
    my_distance = minkowski_distance(A, B, p)
    scipy_distance = minkowski(A, B, p)
    print(p, "\t", round(my_distance, 4), "\t\t", round(scipy_distance, 4))
```

```python
''''a7'''
import numpy as np
def dot_product(A, B):
    dot = 0

    for i in range(len(A)):
        dot = dot + (A[i] * B[i])

    return dot

def euclidean_norm(A):

    total = 0

    for value in A:
        total = total + value ** 2

    return total ** 0.5
print("Dot Product")
print("My Function =", dot_product(A, B))
print("Numpy =", np.dot(A, B))
print()
print("Euclidean Norm of A")
print("My Function =", euclidean_norm(A))
print("Numpy =", np.linalg.norm(A))
print()
print("Euclidean Norm of B")
print("My Function =", euclidean_norm(B))
print("Numpy =", np.linalg.norm(B))
```

```python
''''a8'''
def mean(data):

    total = 0

    for value in data:
        total = total + value

    return total / len(data)
def variance(data):

    avg = mean(data)

    total = 0

    for value in data:
        total = total + (value - avg) ** 2

    return total / len(data)

def standard_deviation(data):
    return variance(data) ** 0.5
features = ["Income", "Kidhome", "Teenhome"]
for feature in features:
    values = []
    for value in df[feature]:
        values.append(value)
    print(feature)
    print("Mean =", mean(values))
    print("Variance =", variance(values))
    print("Standard Deviation =", standard_deviation(values))
    print()
```
